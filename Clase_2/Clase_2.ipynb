```python
# Ejercicio 2 - Tarea 1: Cargar el dataset Automobile.csv
df_auto = pd.read_csv('Automobile.csv')

print(f"Dimensiones del dataset: {df_auto.shape[0]} filas y {df_auto.shape[1]} columnas")
display(df_auto.head())
```

```python
# Tarea 2: Gráfico de dispersión para analizar la relación entre consumo (mpg) y peso (weight)
plt.figure(figsize=(8, 5))
plt.scatter(df_auto['weight'], df_auto['mpg'], color='#1f77b4', alpha=0.7, edgecolors='black')

plt.title("Relación entre Peso del Vehículo y Rendimiento (mpg)", fontsize=13)
plt.xlabel("Peso del automóvil (weight)", fontsize=11)
plt.ylabel("Millas por galón (mpg)", fontsize=11)
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
# Tarea 3: Histograma del consumo de combustible (mpg) para observar su distribución
plt.figure(figsize=(8, 5))
plt.hist(df_auto['mpg'].dropna(), bins=15, color='#2ca02c', edgecolor='black', alpha=0.8)

plt.title("Distribución del Consumo de Combustible (mpg)", fontsize=13)
plt.xlabel("Millas por galón (mpg)", fontsize=11)
plt.ylabel("Frecuencia (Cantidad de vehículos)", fontsize=11)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
print("Nombres exactos de las columnas:", list(df_auto.columns))
print("Tipos de datos:\n", df_auto.dtypes)
```

```python
# Tarea 4: Diagrama de caja para 'cylinders' (y 'origin' si es numérica)
cols_numericas = df_auto.select_dtypes(include=['number']).columns

# Buscar las columnas de la consigna sin importar mayúsculas
col_cylinders = [c for c in df_auto.columns if 'cyl' in c.lower()]
col_origin = [c for c in df_auto.columns if 'orig' in c.lower()]

columnas_a_graficar = []
if col_cylinders and col_cylinders[0] in cols_numericas:
    columnas_a_graficar.append(col_cylinders[0])
if col_origin and col_origin[0] in cols_numericas:
    columnas_a_graficar.append(col_origin[0])

# Si origin es texto, graficamos cylinders que es la variable con atípicos
if not columnas_a_graficar and col_cylinders:
    columnas_a_graficar = [col_cylinders[0]]

plt.figure(figsize=(8, 5))
datos = [df_auto[c].dropna() for c in columnas_a_graficar]
plt.boxplot(datos, labels=columnas_a_graficar, patch_artist=True,
            boxprops=dict(facecolor='#aec7e8', color='navy'),
            medianprops=dict(color='red', linewidth=2))

plt.title("Diagrama de Caja para Identificar Valores Atípicos", fontsize=13)
plt.ylabel("Valores", fontsize=11)
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
# Tarea 5: Distribución según año de modelo y origen
col_year = [c for c in df_auto.columns if 'year' in c.lower() or 'model' in c.lower()][0]
col_orig = [c for c in df_auto.columns if 'orig' in c.lower()][0]

conteo = df_auto.groupby([col_year, col_orig]).size().unstack(fill_value=0)

conteo.plot(kind='bar', stacked=True, figsize=(10, 5), colormap='Set2', edgecolor='black')
plt.title("Distribución de Automóviles por Año de Modelo y Origen", fontsize=13)
plt.xlabel("Año del Modelo", fontsize=11)
plt.ylabel("Cantidad de Automóviles", fontsize=11)
plt.legend(title="Origen")
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
